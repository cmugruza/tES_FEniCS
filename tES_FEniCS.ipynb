```
#2025.03.12 Interneurons
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Hodgkin-Huxley Model for Neurons
class HodgkinHuxley:
    C_m = 1.0  # membrane capacitance, in uF/cm^2
    g_Na = 120.0  # maximum conductances, in mS/cm^2
    g_K = 36.0
    g_L = 0.3
    E_Na = 50.0  # Nernst reversal potentials, in mV
    E_K = -77.0
    E_L = -54.387

    def __init__(self, condition=None):
        if condition == "Alzheimer":
            self.g_Na *= 1.5  # Example adjustment for Alzheimer's
            self.g_K *= 0.8   # Example adjustment for Alzheimer's
        elif condition == "Stroke":
            self.g_Na *= 1.2  # Example adjustment for Stroke
            self.g_K *= 1.2   # Example adjustment for Stroke
        elif condition == "Anemia":
            self.g_Na *= 0.9  # Example adjustment for Anemia
            self.g_K *= 1.1   # Example adjustment for Anemia

    def alpha_m(self, V): return 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))
    def beta_m(self, V): return 4.0 * np.exp(-(V + 65) / 18)
    def alpha_h(self, V): return 0.07 * np.exp(-(V + 65) / 20)
    def beta_h(self, V): return 1.0 / (1 + np.exp(-(V + 35) / 10))
    def alpha_n(self, V): return 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))
    def beta_n(self, V): return 0.125 * np.exp(-(V + 65) / 80)

    def __call__(self, X, t, I):
        V, m, h, n = X
        dVdt = (I - self.g_Na * m**3 * h * (V - self.E_Na) - self.g_K * n**4 * (V - self.E_K) - self.g_L * (V - self.E_L)) / self.C_m
        dmdt = self.alpha_m(V) * (1 - m) - self.beta_m(V) * m
        dhdt = self.alpha_h(V) * (1 - h) - self.beta_h(V) * h
        dndt = self.alpha_n(V) * (1 - n) - self.beta_n(V) * n
        return dVdt, dmdt, dhdt, dndt

# Define parameters
frequencies = [1, 10, 100, 1000]  # Hz (only for tACS)
intensities = [100, 150, 200]     # μA
conductivity = 0.33               # S/m (typical brain tissue conductivity)
grid_size = 100                   # Grid size for simulation (100x100)
electrode_distance = 10           # Distance between electrodes in mm

# Simulation Parameters
TREATMENT = "tACS" # Choose "tDCS" or "tACS"

# Time-stepping parameters (used in tACS)
time_step = 0.1  # Time step in milliseconds
total_time = 50  # Total simulation time in milliseconds

# Create a grid
x = np.linspace(-electrode_distance/2, electrode_distance/2, grid_size)
y = np.linspace(-electrode_distance/2, electrode_distance/2, grid_size)
X, Y = np.meshgrid(x, y)

# Function to compute electric potential (tDCS)
def compute_potential_tDCS(intensity):
    V = intensity / (2 * np.pi * conductivity * np.sqrt(X**2 + Y**2 + 1e-9))
    return V

# Function to compute electric potential (tACS)
def compute_potential_tACS(intensity, frequency, time):
    V = intensity * np.sin(2 * np.pi * frequency * time/1000) / (2 * np.pi * conductivity * np.sqrt(X**2 + Y**2 + 1e-9))
    return V

# Simulate firing rates
def simulate_firing_rate(model, I_ext, t_max=50):
    t = np.linspace(0, t_max, 1000)
    X0 = [-65, 0.05, 0.6, 0.32]
    sol = odeint(model, X0, t, args=(I_ext,))
    V = sol[:, 0]
    spikes = np.where((V[:-1] < 0) & (V[1:] >= 0))[0]
    firing_rate = len(spikes) / t_max * 1000  # spikes per second
    return firing_rate

# Plot the results
fig, axes = plt.subplots(len(frequencies), len(intensities), figsize=(15, 15))
fig.suptitle(f'Electric Fields and Firing Rates (Treatment = {TREATMENT})', fontsize=16)

hh_model_pyramidal = HodgkinHuxley()
hh_model_interneuron = HodgkinHuxley()  # Assume similar model for simplicity

if TREATMENT == "tDCS":
    # Simulate tDCS
    for i, intensity in enumerate(intensities):  # only intensities are relevant for tDCS
            V = compute_potential_tDCS(intensity)

            E_x, E_y = np.gradient(-V)  # Electric field components
            E_magnitude = np.sqrt(E_x**2 + E_y**2)

            # Compute firing rates for pyramidal and interneurons
            I_ext_pyramidal = E_magnitude.mean()  # Simplified external current for simulation
            I_ext_interneuron = E_magnitude.mean() * 1.01

            firing_rate_pyramidal = simulate_firing_rate(hh_model_pyramidal, I_ext_pyramidal)
            firing_rate_interneuron = simulate_firing_rate(hh_model_interneuron, I_ext_interneuron)

            #Plotting tDCS needs some modification

            ax = axes[0, i] # for plotting purposes only
            c = ax.contourf(X, Y, E_magnitude, levels=20, cmap='viridis')
            ax.set_title(f'{intensity} μA\nPyramidal: {firing_rate_pyramidal:.2f} Hz, Interneuron: {firing_rate_interneuron:.2f} Hz')
            fig.colorbar(c, ax=ax)
            ax.set_xlabel('X (mm)')
            ax.set_ylabel('Y (mm)')
else:  # TREATMENT == "tACS"
    # Simulate tACS
    for i, frequency in enumerate(frequencies):
        for j, intensity in enumerate(intensities):
            #V = compute_potential_tACS(intensity, frequency, t)
            time = np.arange(0, total_time, time_step) # array of time steps
            firing_rate_pyramidal = []
            firing_rate_interneuron = []

            for t in time:
                V = compute_potential_tACS(intensity, frequency, t)
                E_x, E_y = np.gradient(-V)  # Electric field components
                E_magnitude = np.sqrt(E_x**2 + E_y**2)
                # Compute firing rates for pyramidal and interneurons

                I_ext_pyramidal = E_magnitude.mean()  # Simplified external current for simulation
                I_ext_interneuron = E_magnitude.mean() * 0.91
                firing_rate_pyramidal.append(simulate_firing_rate(hh_model_pyramidal, I_ext_pyramidal))
                firing_rate_interneuron.append(simulate_firing_rate(hh_model_interneuron, I_ext_interneuron))
            #calculate mean firing rate over time array
            firing_rate_pyramidal = np.mean(firing_rate_pyramidal)
            firing_rate_interneuron = np.mean(firing_rate_interneuron)
            ax = axes[i, j]
            c = ax.contourf(X, Y, E_magnitude, levels=20, cmap='viridis')
            ax.set_title(f'{frequency} Hz, {intensity} μA\nPyramidal: {firing_rate_pyramidal:.2f} Hz, Interneuron: {firing_rate_interneuron:.2f} Hz')
            fig.colorbar(c, ax=ax)
            ax.set_xlabel('X (mm)')
            ax.set_ylabel('Y (mm)')

plt.tight_layout()
plt.show()
```

```
# Corrected simulation, changing number of time steps to 10000 and ploting in mm
# 2025.03.11
import numpy as np
import matplotlib.pyplot as plt
from scipy.sparse import diags
from scipy.sparse.linalg import spsolve
import time

# Define parameters
nx, ny = 50, 50
Lx, Ly = 0.1, 0.1  # Domain size (10 cm x 10 cm)
dx, dy = Lx / nx, Ly / ny
frequency = 14.0  # tACS frequency in Hz
amplitude = 1.0   # Amplitude of the sinusoidal potential

# Time parameters
start_time = 0.0    # Start time (s)
end_time = 1.0      # End time (s)
time_step = 0.1e-3   # Time step (0.1 ms)
time_points = np.arange(start_time, end_time, time_step)  # Time points array

conductivity = 0.9  # Conductivity of the medium (S/m)

# Electrode dimensions and locations
electrode_width = 0.02  # Electrode width (2 cm)
electrode_height = 0.01  # Electrode height (1 cm)
electrode_gap = 0.06    # Distance between electrode centers (6 cm)

# Create grid
x = np.linspace(0, Lx, nx)
y = np.linspace(0, Ly, ny)
X, Y = np.meshgrid(x, y)

# Define electrode mask function
def electrode_mask(X, Y, center_x, width, height):
    return ((X >= center_x - width/2) & (X <= center_x + width/2) &
            (Y >= 0) & (Y <= height))

# Electrode centers
anode_center_x = Lx / 2 - electrode_gap / 2
cathode_center_x = Lx / 2 + electrode_gap / 2

# Create electrode masks
anode = electrode_mask(X, Y, anode_center_x, electrode_width, electrode_height)
cathode = electrode_mask(X, Y, cathode_center_x, electrode_width, electrode_height)

# Initialize potential array
V = np.zeros((nx, ny))

# Set up the Laplacian matrix with conductivity
main_diag = 4 * np.ones(nx * ny) * conductivity
off_diag_x = -1 * np.ones(nx * ny) * conductivity
off_diag_y = -1 * np.ones(nx * ny) * conductivity

A = diags([off_diag_x[:-1], off_diag_y[:-nx], main_diag, off_diag_y[:-nx], off_diag_x[:-1]],
          [-1, -nx, 0, nx, 1], shape=(nx*ny, nx*ny)).tocsc()

# Prepare to store potential and electric field at different time points
potential_snapshots = []
electric_field_snapshots = []

# Simulate over time
for t in time_points:
    # Apply sinusoidal potential at electrodes
    anode_potential = amplitude * np.sin(2 * np.pi * frequency * t)
    V[anode] = anode_potential
    V[cathode] = 0  # Grounded cathode

    # Boundary conditions (Dirichlet)
    b = np.zeros(nx * ny)
    for i in range(nx):
        for j in range(ny):
            index = i * ny + j
            if anode[i, j]:
                A[index, :] = 0
                A[index, index] = 1
                b[index] = anode_potential
            elif cathode[i, j]:
                A[index, :] = 0
                A[index, index] = 1
                b[index] = 0

    # Solve the system
    V_flat = spsolve(A, b)
    V = V_flat.reshape((nx, ny))

    # Calculate electric field
    Ex, Ey = np.gradient(-V, dx, dy)

    # Store snapshot of potential and electric field
    potential_snapshots.append(V.copy())
    electric_field_snapshots.append((Ex.copy(), Ey.copy()))

# Choose a time point to visualize (e.g., the last one)
time_idx = -1 #Visualize last time point (t=1)

# Plot the results at a specific time point
V = potential_snapshots[time_idx]
Ex, Ey = electric_field_snapshots[time_idx]

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.contourf(X * 1000, Y * 1000, V, 20, cmap='viridis')
plt.colorbar(label='Electric Potential (V)')
plt.xlabel('x (mm)')
plt.ylabel('y (mm)')
plt.title(f'Electric Potential Distribution at t = {time_points[time_idx]:.3f} s')

# Plot electric field
plt.subplot(1, 2, 2)
plt.quiver(X * 1000, Y * 1000, Ex, Ey)
plt.xlabel('x (mm)')
plt.ylabel('y (mm)')
plt.title(f'Electric Field (V/m) at t = {time_points[time_idx]:.3f} s')

plt.tight_layout()
plt.show()
```

```
# WORKING 2025.02.18
"""
Python implementation of the Hodgkin-Huxley spiking neuron model
https://github.com/swharden/pyHH
"""
import matplotlib.pyplot as plt
import numpy as np


class HHModel:
    """The HHModel tracks conductances of 3 channels to calculate Vm"""

    class Gate:
        """The Gate object manages a channel's kinetics and open state"""
        alpha, beta, state = 0, 0, 0

        def update(self, deltaTms):
            alphaState = self.alpha * (1-self.state)
            betaState = self.beta * self.state
            self.state += deltaTms * (alphaState - betaState)

        def setInfiniteState(self):
            self.state = self.alpha / (self.alpha + self.beta)

    ENa, EK, EKleak = 115, -12, 10.6
    gNa, gK, gKleak = 120, 36, 0.3
    #gNa, gK = 116.33, 25.64  # Example adjustment for Anemia 120.0*131.42/135.57 = 116.33
                            # Example adjustment for Anemia 36.0*4.37/4.09*2/3 = 25.64
    m, n, h = Gate(), Gate(), Gate()
    Cm = 1

    def __init__(self, startingVoltage=0):
        self.Vm = startingVoltage
        self.UpdateGateTimeConstants(startingVoltage)
        self.m.setInfiniteState()
        self.n.setInfiniteState()
        self.n.setInfiniteState()

    def UpdateGateTimeConstants(self, Vm):
        """Update time constants of all gates based on the given Vm"""
        self.n.alpha = .01 * ((10-Vm) / (np.exp((10-Vm)/10)-1))
        self.n.beta = .125*np.exp(-Vm/80)
        self.m.alpha = .1*((25-Vm) / (np.exp((25-Vm)/10)-1))
        self.m.beta = 4*np.exp(-Vm/18)
        self.h.alpha = .07*np.exp(-Vm/20)
        self.h.beta = 1/(np.exp((30-Vm)/10)+1)

    def UpdateCellVoltage(self, stimulusCurrent, deltaTms):
        """calculate channel currents using the latest gate time constants"""
        INa = np.power(self.m.state, 3) * self.gNa * \
            self.h.state*(self.Vm-self.ENa)
        IK = np.power(self.n.state, 4) * self.gK * (self.Vm-self.EK)
        IKleak = self.gKleak * (self.Vm-self.EKleak)
        Isum = stimulusCurrent - INa - IK - IKleak
        self.Vm += deltaTms * Isum / self.Cm

    def UpdateGateStates(self, deltaTms):
        """calculate new channel open states using latest Vm"""
        self.n.update(deltaTms)
        self.m.update(deltaTms)
        self.h.update(deltaTms)

    def Iterate(self, stimulusCurrent=0, deltaTms=0.05):
        self.UpdateGateTimeConstants(self.Vm)
        self.UpdateCellVoltage(stimulusCurrent, deltaTms)
        self.UpdateGateStates(deltaTms)


if __name__ == "__main__":
    hh = HHModel()
    pointCount = 20000
    frequency = 6
    tES = 2
    voltages = np.empty(pointCount)
    times = np.arange(pointCount) * 0.05
    stim = np.zeros(pointCount)
    if tES == 1:
      stim[1000:pointCount-1000] = 20  # create a square tDCS pulse
      text_figure = f"tDCS"
    if tES == 2:
      stim[1000:pointCount-1000] = 5*np.sin(2 * np.pi * frequency/1000 * times[1000:pointCount-1000])  # create a tACS pulse
      text_figure = f" tACS @ {frequency} Hz"

#def oscillating_current(t, frequency, intensity):
#    return intensity * np.sin(2 * np.pi * frequency * times)
# Define parameters for tES simulation
#frequencies_tES = [1, 10, 100]   # tES frequencies in Hz to test
#intensity_tES = [250.0e-6]          # μA
    for i in range(len(times)):
        hh.Iterate(stimulusCurrent=stim[i], deltaTms=0.01) #deltaTms .05 to .01 to simulate lower frequencies
        voltages[i] = hh.Vm
        # note: you could also plot hh's n, m, and k (channel open states)

    f, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 5),
                                 gridspec_kw={'height_ratios': [3, 1]})

    ax1.plot(times, voltages - 70, 'b')
    ax1.set_ylabel("Membrane Potential (mV)")
    ax1.set_title(f"Hodgkin-Huxley Spiking Neuron Model at {text_figure}", fontsize=16)
    ax1.spines['right'].set_visible(False)
    ax1.spines['top'].set_visible(False)
    ax1.spines['bottom'].set_visible(False)
    ax1.tick_params(bottom=False)

    ax2.plot(times, stim, 'r')
    ax2.set_ylabel("Stimulus (µA/cm²)")
    ax2.set_xlabel("Simulation Time (ms)")
    ax2.spines['right'].set_visible(False)
    ax2.spines['top'].set_visible(False)

    plt.margins(0, 0.1)
    plt.tight_layout()
    plt.savefig("concept4.png")
    plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
frequencies = [1, 10, 100, 1000]  # Hz
intensities = [100, 150, 200]     # μA
conductivity = 0.33               # S/m (typical brain tissue conductivity)
grid_size = 100                   # Grid size for simulation (100x100)
electrode_distance = 17.81        # Distance between electrodes in mm

# Hodgkin-Huxley parameters for pyramidal neurons
V_rest = -65.0  # Resting membrane potential in mV
g_Na = 120.0    # Maximum conductances in mS/cm^2
g_K = 36.0
g_L = 0.3
E_Na = 50.0     # Nernst reversal potentials in mV
E_K = -77.0
E_L = -54.387

# Hodgkin-Huxley parameters for interneurons (simplified)
V_rest_interneuron = -65.0
g_Na_interneuron = 120.0
g_K_interneuron = 36.0
g_L_interneuron = 0.3
E_Na_interneuron = 50.0
E_K_interneuron = -77.0
E_L_interneuron = -54.387

# Simulation parameters
dt = 0.01  # Time step in ms
simulation_time = 100  # Total simulation time in ms

# Create a grid
x = np.linspace(-electrode_distance/2, electrode_distance/2, grid_size)
y = np.linspace(-electrode_distance/2, electrode_distance/2, grid_size)
X, Y = np.meshgrid(x, y)

# Function to compute electric potential
def compute_potential(intensity, frequency):
    V = intensity / (2 * np.pi * conductivity * np.sqrt(X**2 + Y**2 + 1e-9))
    return V

# Hodgkin-Huxley model for pyramidal neurons
def compute_firing_rate(V_ext):
    V = V_rest
    m = h = n = 0.5  # Initial values for gating variables
    firing_rate = 0
    for t in np.arange(0, simulation_time, dt):
        alpha_m = 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))
        beta_m = 4.0 * np.exp(-(V + 65) / 18)
        alpha_h = 0.07 * np.exp(-(V + 65) / 20)
        beta_h = 1.0 / (1 + np.exp(-(V + 35) / 10))
        alpha_n = 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))
        beta_n = 0.125 * np.exp(-(V + 65) / 80)

        m += dt * (alpha_m * (1 - m) - beta_m * m)
        h += dt * (alpha_h * (1 - h) - beta_h * h)
        n += dt * (alpha_n * (1 - n) - beta_n * n)

        I_Na = g_Na * m**3 * h * (V - E_Na)
        I_K = g_K * n**4 * (V - E_K)
        I_L = g_L * (V - E_L)

        V += dt * (-I_Na - I_K - I_L + V_ext[int(len(V_ext)/2)][int(len(V_ext)/2)])

        if V >= 0:
            firing_rate += 1
            V = V_rest  # Reset after spike

    return firing_rate

# Hodgkin-Huxley model for interneurons (simplified)
def compute_firing_rate_interneuron(V_ext):
    V = V_rest_interneuron
    m = h = n = 0.5
    firing_rate = 0
    for t in np.arange(0, simulation_time, dt):
        alpha_m = 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))
        beta_m = 4.0 * np.exp(-(V + 65) / 18)
        alpha_h = 0.07 * np.exp(-(V + 65) / 20)
        beta_h = 1.0 / (1 + np.exp(-(V + 35) / 10))
        alpha_n = 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))
        beta_n = 0.125 * np.exp(-(V + 65) / 80)

        m += dt * (alpha_m * (1 - m) - beta_m * m)
        h += dt * (alpha_h * (1 - h) - beta_h * h)
        n += dt * (alpha_n * (1 - n) - beta_n * n)

        I_Na = g_Na_interneuron * m**3 * h * (V - E_Na_interneuron)
        I_K = g_K_interneuron * n**4 * (V - E_K_interneuron)
        I_L = g_L_interneuron * (V - E_L_interneuron)

        V += dt * (-I_Na - I_K - I_L + V_ext[int(len(V_ext)/2)][int(len(V_ext)/2)])

        if V >= 0:
            firing_rate += 1
            V = V_rest_interneuron  # Reset after spike

    return firing_rate

# Plot the results
fig, axes = plt.subplots(len(frequencies), len(intensities), figsize=(15, 15))
fig.suptitle('Electric Fields and Firing Rates of Pyramidal and Interneuron Neurons', fontsize=16)

for i, frequency in enumerate(frequencies):
    for j, intensity in enumerate(intensities):
        V_ext = compute_potential(intensity, frequency)
        firing_rate_pyramidal = compute_firing_rate(V_ext)
        firing_rate_interneuron = compute_firing_rate_interneuron(V_ext)

        ax = axes[i, j]
        c = ax.contourf(X, Y, V_ext, levels=20, cmap='viridis')
        ax.set_title(f'{frequency} Hz, {intensity} μA\nPyramidal: {firing_rate_pyramidal:.2f} Hz, Interneuron: {firing_rate_interneuron:.2f} Hz')
        fig.colorbar(c, ax=ax)
        ax.set_xlabel('X (mm)')
        ax.set_ylabel('Y (mm)')

plt.tight_layout()
plt.show()
```
